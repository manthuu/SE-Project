```python

import pandas as pd

# Keep only the useful columns
required_columns = [
    "source_article",
    "updated_label",
    "masked_articles"
]

train_df = train_df[required_columns].copy()
dev_df = dev_df[required_columns].copy()
test_df = test_df[required_columns].copy()

# Remove rows with missing text or labels, if any
for df in [train_df, dev_df, test_df]:
    df.dropna(subset=["source_article", "updated_label"], inplace=True)
    df["source_article"] = df["source_article"].astype(str).str.strip()
    df["updated_label"] = df["updated_label"].astype(str).str.strip()
    df.drop(
        df[(df["source_article"] == "") |
           (df["updated_label"] == "")].index,
        inplace=True
    )

# Display the cleaned data
print("Cleaned dataset sizes:")
print("Training:", train_df.shape)
print("Validation:", dev_df.shape)
print("Testing:", test_df.shape)

print("\nCleaned columns:")
print(train_df.columns.tolist())

display(train_df.head())
```

```python
import torch

premise = "Everyone is buying this product, so it must be the best."
hypothesis = "This is an example of ad populum logical fallacy"

inputs = tokenizer(
    premise,
    hypothesis,
    return_tensors="pt",
    truncation=True,
    max_length=512
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)
inputs = {k: v.to(device) for k, v in inputs.items()}

model.eval()
with torch.no_grad():
    outputs = model(**inputs)
    probabilities = torch.softmax(outputs.logits, dim=-1)[0]

labels = ["Entailment", "Contradiction", "Neutral"]

print("Device:", device)
for label, score in zip(labels, probabilities.tolist()):
    print(f"{label}: {score:.4f}")
```

```python
import torch
import pandas as pd

argument = "Everyone is buying this product, so it must be the best."

categories = sorted(train_df["updated_label"].unique())
results = []

model.eval()

for category in categories:
    hypothesis = f"This is an example of {category} logical fallacy"

    inputs = tokenizer(
        argument,
        hypothesis,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        logits = model(**inputs).logits
        probs = torch.softmax(logits, dim=-1)[0].cpu().tolist()

    results.append({
        "Fallacy": category,
        "Entailment": round(probs[0], 4),
        "Contradiction": round(probs[1], 4),
        "Neutral": round(probs[2], 4)
    })

results_df = pd.DataFrame(results)
display(results_df.sort_values("Entailment", ascending=False))
```

```python

import torch

FALLACY_CATEGORIES = [
    "ad hominem",
    "ad populum",
    "appeal to emotion",
    "circular reasoning",
    "equivocation",
    "fallacy of credibility",
    "fallacy of extension",
    "fallacy of logic",
    "fallacy of relevance",
    "false causality",
    "false dilemma",
    "faulty generalization",
    "intentional"
]

def predict_fallacies(argument):
    argument = argument.strip()

    if not argument:
        return []

    results = []

    model.eval()

    with torch.no_grad():
        for category in FALLACY_CATEGORIES:
            hypothesis = (
                f"This is an example of {category} logical fallacy"
            )

            inputs = tokenizer(
                argument,
                hypothesis,
                return_tensors="pt",
                truncation=True,
                max_length=256
            )

            outputs = model(**inputs)
            scores = torch.softmax(outputs.logits, dim=-1)[0]

            results.append({
                "fallacy": category,
                "entailment_score": float(scores[0]),
                "contradiction_score": float(scores[1]),
                "neutral_score": float(scores[2])
            })

    results.sort(
        key=lambda item: item["entailment_score"],
        reverse=True
    )

    return results


# Test all 13 categories
test_argument = (
    "Everyone believes this product is the best, "
    "so it must be the best product."
)

predictions = predict_fallacies(test_argument)

for index, result in enumerate(predictions[:5], start=1):
    print(
        f"{index}. {result['fallacy']}: "
        f"{result['entailment_score']:.4f}"
    )
```

```python

import torch
import torch.nn.functional as F

FALLACY_CATEGORIES = [
    "ad hominem",
    "ad populum",
    "appeal to emotion",
    "circular reasoning",
    "equivocation",
    "fallacy of credibility",
    "fallacy of extension",
    "fallacy of logic",
    "fallacy of relevance",
    "false causality",
    "false dilemma",
    "faulty generalization",
    "intentional"
]

def predict_fallacies(argument, top_k=5):
    # Validate the input
    if not isinstance(argument, str) or not argument.strip():
        raise ValueError("Please enter a non-empty argument.")

    if not isinstance(top_k, int) or not 1 <= top_k <= len(FALLACY_CATEGORIES):
        raise ValueError("top_k must be between 1 and 13.")

    model.eval()
    device = next(model.parameters()).device

    # Compare the argument with each fallacy hypothesis
    hypotheses = [
        f"This is an example of {category} logical fallacy."
        for category in FALLACY_CATEGORIES
    ]

    encoded = tokenizer(
        [argument] * len(hypotheses),
        hypotheses,
        padding=True,
        truncation=True,
        max_length=256,
        return_tensors="pt"
    )

    encoded = {key: value.to(device) for key, value in encoded.items()}

    with torch.no_grad():
        outputs = model(**encoded)
        probabilities = F.softmax(outputs.logits, dim=-1)

    # The project's model uses NLI labels:
    # entailment = 0, contradiction = 1, neutral = 2
    entailment_scores = probabilities[:, 0]

    ranked_indices = torch.argsort(
        entailment_scores, descending=True
    )[:top_k]

    results = []
    for index in ranked_indices.tolist():
        results.append({
            "fallacy": FALLACY_CATEGORIES[index],
            "score": round(entailment_scores[index].item(), 4)
        })

    return results

print("predict_fallacies() defined successfully!")
```
